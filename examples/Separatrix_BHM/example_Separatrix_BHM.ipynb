```python
%load_ext autoreload
%autoreload 2

import copy
import os
import random
import glob
import pandas as pd

import holoviews as hv
from holoviews import opts
hv.extension('bokeh') # matplotlib
#hv.output(fig='svg')

from calibtool.CalibManager import CalibManager
from calibtool.algorithms.Separatrix_BHM import Separatrix_BHM
from calibtool.plotters.SeparatrixBHMPlotter import SeparatrixBHMPlotter
from dtk.utils.core.DTKConfigBuilder import DTKConfigBuilder
from simtools.SetupParser import SetupParser

from dtk.utils.builders.TemplateHelper import TemplateHelper
from dtk.utils.builders.ConfigTemplate import ConfigTemplate
from dtk.utils.builders.TaggedTemplate import CampaignTemplate
from dtk.utils.builders.TaggedTemplate import DemographicsTemplate
from simtools.Utilities import verbose_timedelta

from EradicationSite import EradicationSite

from wand.image import Image as WImage
from IPython.display import display

from IPython.display import clear_output
import matplotlib.cm as cm
import matplotlib.colors as mpl_colors
from datetime import datetime
import numpy as np
from IPython.utils import io
```

### Separatrix Config Params

```python
MAX_ITERATIONS = 6 # Number of iterations
IMPLAUSIBILITY_THRESHOLD = 3
TARGET_SUCCESS_FRACTION = 0.5
NUM_PAST_ITERATIONS_TO_INCLUDE_IN_METAMODEL = -1
SAMPLES_PER_ITERATION = 32
SAMPLES_FINAL_ITERATION = 128
TRAINING_FRAC = 0.8
```

### Parameters to be calibrated

```python
params = [
    {
        'Name': 'Base Infectivity',
        #'MapTo': 'Base_Infectivity', # <-- DEMO: Custom mapping, see map_sample_to_model_input below
        'Min': 0,
        'Max': 0.4
    },
    {
        'Name': 'Vaccine Coverage',
        'MapTo': 'Demographic_Coverage__KP_Vaccine_Coverage',
        'Min': 0,
        'Max': 1.0
    },
]
```

### Configure builder using templates

```python
# Which simtools.ini block to use for this calibration
SetupParser.default_block = 'HPC'

# List of sites we want to analyze, you can list more than one, e.g. NdiopCalibSite()
sites = [ EradicationSite() ]

# Choose how you'd like to plot, you can use the SeparatrixPlotter.
plotters = [
    SeparatrixBHMPlotter()  # SeparatrixPlotter must be last because it calls gc.collect()
]

dir_path = os.getcwd()

demog = DemographicsTemplate.from_file( 'demographics.json' )
cfg = ConfigTemplate.from_file( 'config.json' )
cpn = CampaignTemplate.from_file( 'campaign.json' )

# Set static parameters, these are one-time overrides to the baseline files
static_params = { 'Simulation_Duration': 90 }
cfg.set_params( static_params )

# Prepare templates
templates = TemplateHelper()
table_base = {
    'ACTIVE_TEMPLATES': [cfg, cpn],
    'TAGS': { 'Separatrix': None }
}

cb = DTKConfigBuilder()
cb.set_input_files_root(dir_path)
cb.ignore_missing = True
```

```python
def constrain_sample(sample):
    """
    This function is called on every samples and allow the user to edit them before they are passed
    to the map_sample_to_model_input function.
    It is useful to round some parameters as demonstrated below.
    Can do much more here, e.g. for
    # Clinical Fever Threshold High <  MSP1 Merozoite Kill Fraction
    if 'Clinical Fever Threshold High' and 'MSP1 Merozoite Kill Fraction' in sample:
        sample['Clinical Fever Threshold High'] = \
            min( sample['Clinical Fever Threshold High'], sample['MSP1 Merozoite Kill Fraction'] )
    You can omit this function by not specifying it in the Separatrix constructor.
    :param sample: The sample coming from the next point algorithm
    :return: The sample with constrained values
    """
    # Convert Falciparum MSP Variants to nearest integer
    if 'Some_Parameter' in sample:
        sample['Some_Parameter'] = int(round(sample['Some_Parameter']))

    return sample
```

```python
def map_sample_to_model_input(cb, sample):
    """
    This method needs to map the samples generated by the next point algorithm to the model inputs (represented here by the cb).
    It is important to note that the sample may be shared by several isntances of this function.
    Therefore it is important to deepcopy the sample at the beginning if we intend to modify it (by calling .pop() for example).
       sample = copy.deepcopy(sample)
    :param cb: The config builder representing the model inputs for this particular simulation
    :param sample: The sample containing a values for all the params. e.g. {'Vaccine Coverage': 0.7, ... }
    :return: A dictionary containing the tags that will be attached to the simulation
    """
    table = copy.deepcopy(table_base)
    table['Run_Number'] = random.randint(0, 1e6) # Random random number seed
    table['Config_Name'] = 'SIR Separatirx'

    # Make a copy of samples so we can alter it safely
    sample = copy.deepcopy(sample)

    # Can perform custom mapping, e.g. a trivial example
    if 'Base Infectivity' in sample:
        value = sample.pop('Base Infectivity')
        table['Base_Infectivity'] = value # Could do value**2 or whatever

    for p in params:
        if 'MapTo' in p:
            value = sample.pop( p['Name'] )
            if isinstance( p['MapTo'], list):
                for mapto in p['MapTo']:
                    table[mapto] = value
            else:
                table[p['MapTo']] = value

    for name,value in sample.items():
        print('UNUSED PARAMETER:'+name)

    assert( len(sample) == 0 ) # All params used

    return templates.mod_dynamic_parameters(cb, table)
```

```python
separatrix = Separatrix_BHM(params,
    constrain_sample,   # <-- WILL NOT BE SAVED IN ITERATION STATE
    implausibility_threshold = IMPLAUSIBILITY_THRESHOLD, # <-- 
    target_success_probability = TARGET_SUCCESS_FRACTION, # <-- 
    num_past_iterations_to_include_in_metamodel = NUM_PAST_ITERATIONS_TO_INCLUDE_IN_METAMODEL, # <-- Number of past iterations to include when emulating.  Set negative to use all.
    samples_per_iteration = SAMPLES_PER_ITERATION, #  <-- Number of samples per iteration.  Actual number of sims run is this number times number of sites.
    samples_final_iteration = SAMPLES_FINAL_ITERATION, #  <-- Number of samples for the final iteration.  Actual number of sims run is this number times number of sites.
    max_iterations = MAX_ITERATIONS,          # <-- Iterations
    training_frac = TRAINING_FRAC
)
```

```python
calib_manager = CalibManager(name='Example_Separatrix_NB',    # <-- Please customize this name
                             config_builder = cb,
                             map_sample_to_model_input_fn = map_sample_to_model_input,
                             sites = sites,
                             next_point = separatrix,
                             sim_runs_per_param_set = 1,  # <-- Replicates
                             max_iterations = MAX_ITERATIONS,          # <-- Iterations
                             plotters = plotters)
```

```python
#calib_manager.run_calibration()
from simtools.Utilities.Experiments import validate_exp_name, retrieve_experiment
from simtools.SetupParser import SetupParser

# Check experiment name as early as possible
if not validate_exp_name(calib_manager.name):
    print('Invalid Experiment!')

try:
    SetupParser.init()
except SetupParser.AlreadyInitialized as e:
    pass

calib_manager.location = SetupParser.get('type')
calib_manager.create_calibration(calib_manager.location)
#calib_manager.run_iterations()
```

```python
def files_to_plot(iteration):
    return [
        'EradicationAnalyzer.png',
        'Separatrix_it%d.png'%iteration,
        'Separatrix_Hyperparameters_it%d.png'%iteration,
        'Separatrix_Implausibility_it%d.png'%iteration,
    ]
```

```python
"""
Run iterations in a loop
"""
iteration = 0
my_cmap = cm.get_cmap('RdBu_r')

calib_manager.calibration_start = datetime.now().replace(microsecond=0)

sp = calib_manager.plotters[0] # Should find by class type    
res = sp.prediction_resolution
[x_var, y_var] = [p['Name'] for p in params]
bounds = (separatrix.param_info.loc[x_var]['Min'], separatrix.param_info.loc[y_var]['Min'], \
          separatrix.param_info.loc[x_var]['Max'], separatrix.param_info.loc[y_var]['Max'])
T = np.zeros((calib_manager.max_iterations-1, res, res))


# normal run
for i in range(iteration, calib_manager.max_iterations):
    print('-'*80 + '\n' + '-'*80 + '\nBEGINNING ITERATION %d\n'%i + '-'*80 + '\n' + '-'*80)
    with io.capture_output() as captured:
        calib_manager.current_iteration = calib_manager.create_iteration_state(i)
        calib_manager.current_iteration.run()
        calib_manager.post_iteration()
    
    # Results get saved to disk, so load and display:
    for filename in files_to_plot(i):#glob.glob(os.path.join(calib_manager.current_iteration.iteration_directory, "*.png")):
        filepath = os.path.join(calib_manager.current_iteration.iteration_directory, filename)
        if os.path.isfile(filepath):
            img = WImage(filename=filepath)
            print(filepath)
            display(img)
    
    if i > 0:
        bot = sp.prediction_grid['Max_Implausibility'].min()
        top = sp.prediction_grid['Max_Implausibility'].max()

        frac = (IMPLAUSIBILITY_THRESHOLD-bot)/(top-bot)
        Nlow = int(frac*256)
        my_colors = my_cmap( np.append(np.linspace(0,0.5,Nlow), np.linspace(0.5,1,256-Nlow)) )
        ncm = mpl_colors.ListedColormap(my_colors, name='RB')
        
        T[i-1 :, :] = sp.prediction_grid['Max_Implausibility'].values.reshape(res,res)

        HM = hv.HoloMap(
            {it: hv.Image(T[it,:,:], [sp.x_var, sp.y_var], bounds=bounds) \
             .opts(width=600, height=600, cmap=ncm, colorbar=True, invert_yaxis=True) for it in np.arange(i)},
            kdims=['Iteration']
        ) #, color_levels=levels
        display(HM)


print('DONE - calling finalize')
final = calib_manager.finalize_calibration();
final_df = pd.DataFrame.from_dict(final['final_samples'], orient='columns');
display(final_df)

# Print the calibration finish time
current_time = datetime.now()
calibration_time_elapsed = current_time - calib_manager.calibration_start
print("Calibration done (took %s)" % verbose_timedelta(calibration_time_elapsed))
```
